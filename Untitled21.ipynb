```python
import nltk
from nltk.util import ngrams
nltk.download("punkt")
nltk.download("punkt_tab")
sentence=("AI is simulation of human intelligence by machine",
          "It includes expects system NLP speech recognition and machine vision")
tokens=nltk.word_tokenize(" ".join(sentence))
bigrams=list(ngrams(tokens,2))
print("bigrams:")
for bigram in bigrams:
  print(bigram)
trigrams=list(ngrams(tokens,3))
print("\ntrigrams:")
for trigram in trigrams:
  print(trigram)
```
